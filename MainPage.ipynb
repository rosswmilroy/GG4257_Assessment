```python
import requests
import pandas as pd
import geopandas as gpd

# Let's describe the url, it is usually easier to do it like this, so in the future, you can easily update the URL
sensors = "https://api.glasgow.gov.uk/traffic/v1/movement/sites"
# Making the query to the web server, using the Get method from the requests library 
response = requests.get(sensors)
response
data = response.json()
data
```

Turn data into a geopackage with correct CRS, and longitude and latitude.

```python
# empty lists to store what we need
site_ids = []
from_lats = []
from_longs = []
from_descriptions = []

# Iterate over each entry in the JSON array
for entry in data:
    site_id = entry['siteId']
    from_description = entry['from']['description']
    from_lat = entry['from']['lat']
    from_long = entry['from']['long']
    
    # Append values to lists
    site_ids.append(site_id)
    from_descriptions.append(from_description)
    from_lats.append(from_lat)
    from_longs.append(from_long)
    
# Create a Pandas DataFrame, using the lists.
df = pd.DataFrame({
    'siteId': site_ids,
    'from_description': from_descriptions,
    'from_lat': from_lats,
    'from_long': from_longs,
})
```

```python
gdf_sensors = gpd.GeoDataFrame(df, geometry=gpd.points_from_xy(df['from_long'], df['from_lat']), crs='EPSG:4326')
gdf_sensors.head()
```

```python
import pandas as pd
listings =pd.read_csv("data/listings.csv")
subset_listings = listings[['id', 
                            'neighbourhood_cleansed',
                            'latitude',
                            'longitude',
                            'property_type',
                            'room_type',
                            'bedrooms',
                            'price',
                            'number_of_reviews']]
subset_listings.loc[subset_listings.index, 'price'] = subset_listings['price'].replace('[\$,]', '', regex=True).astype(float)
subset_listings = subset_listings.dropna()
```

```python
# data_description function
def data_description(x):
    mean_array = []
    count_array = []
    name_array = []
    array = ["Name","Mean","Count"]
    counter = 0
    for counter in range(len(x.columns)):
        name_array.append(x.columns[counter])
        mean_array.append(x.iloc[:, (counter)].mean())
        count_array.append(x.iloc[:, (counter)].count())
    dictionary = dict(zip((array),(name_array, mean_array, count_array))) 
    return dictionary

num_subset_listings = subset_listings.select_dtypes(include=["number"])
dictionary = data_description(num_subset_listings)
dataframe = pd.DataFrame(dictionary)
pd.set_option('display.float_format',lambda x: '%.4f' %x)
dataframe

#This was how I learnt to append an array
#https://www.geeksforgeeks.org/python/python-add-array-item/

#This was how I learnt to use counter to cycle through the dataframe to calculate the mean and count
#https://stackoverflow.com/questions/36316154/call-column-in-dataframe-by-column-index-instead-of-column-name-pandas

#This was how I converted my three arrays into a dictionary
#https://www.geeksforgeeks.org/python/python-convert-two-lists-into-a-dictionary/

#For fun, decided I wanted to see if i could sort the output to have a pretty notation and not scientific :)
#https://stackoverflow.com/questions/21137150/format-suppress-scientific-notation-from-pandas-aggregation-results
```
